```python
from collections import deque

# ============================================================
# 8-PUZZLE STATE DEFINTION
# ============================================================

# Simpler initial state (only 2 moves away from GOAL) for small, readable outputs
START = (
    1, 4, 2,
    3, 0, 5,
    6, 7, 8
)

GOAL = (
    0, 1, 2,
    3, 4, 5,
    6, 7, 8
)

# ============================================================
# PRINT THE PUZZLE
# ============================================================

def print_state(state):
    """Print a 3 x 3 puzzle board."""
    for i in range(0, 9, 3):
        row = state[i:i + 3]
        # Display 0 as blank
        print(" ".join("□" if x == 0 else str(x) for x in row))
    print()

# ============================================================
# GENERATE LEGAL MOVES
# ============================================================

def get_neighbors(state):
    """Generate all legal moves of the blank."""
    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    directions = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    neighbors = []
    for dr, dc, move in directions:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank], new_state[new_blank] = new_state[new_blank], new_state[blank]
            neighbors.append((tuple(new_state), move))

    return neighbors

# ============================================================
# RECONSTRUCT SOLUTION PATH
# ============================================================

def reconstruct_path(parent, move_used, goal):
    """Move backwards using parent[] until START is reached."""
    path = []
    current = goal
    while current is not None:
        path.append((current, move_used[current]))
        current = parent[current]
    path.reverse()
    return path
```

```python
# ============================================================
# BFS - BREADTH FIRST SEARCH
# ============================================================

def bfs(start, goal):
    queue = deque()
    queue.append(start)
    visited = {start}
    parent = {start: None}
    move_used = {start: None}

    while queue:
        current = queue.popleft()
        if current == goal:
            return reconstruct_path(parent, move_used, goal)

        for next_state, move in get_neighbors(current):
            if next_state not in visited:
                visited.add(next_state)
                parent[next_state] = current
                move_used[next_state] = move
                queue.append(next_state)
    return None
```

```python
# ============================================================
# DFS - DEPTH FIRST SEARCH
# ============================================================

def dfs(start, goal):
    stack = []
    stack.append(start)
    visited = set()
    parent = {start: None}
    move_used = {start: None}

    while stack:
        current = stack.pop()
        if current in visited:
            continue
        visited.add(current)

        if current == goal:
            return reconstruct_path(parent, move_used, goal)

        neighbors = get_neighbors(current)
        for next_state, move in reversed(neighbors):
            if next_state not in visited:
                if next_state not in parent:
                    parent[next_state] = current
                    move_used[next_state] = move
                stack.append(next_state)
    return None

# Run DFS
dfs_solution = dfs(START, GOAL)
display_solution("DEPTH FIRST SEARCH (DFS)", dfs_solution)
```

```python
# ============================================================
# IDS - ITERATIVE DEEPENING SEARCH & DEPTH LIMITED SEARCH
# ============================================================

def depth_limited_search(current, goal, depth_limit, path, visited):
    if current == goal:
        return path
    if depth_limit == 0:
        return None

    visited.add(current)
    for next_state, move in get_neighbors(current):
        if next_state not in visited:
            path.append((next_state, move))
            result = depth_limited_search(next_state, goal, depth_limit - 1, path, visited)
            if result is not None:
                return result
            path.pop() # Backtrack

    visited.remove(current)
    return None

def ids(start, goal):
    depth_limit = 0
    while True:
        path = [(start, None)]
        visited = set()
        result = depth_limited_search(start, goal, depth_limit, path, visited)
        if result is not None:
            return result
        depth_limit += 1

# Run IDS
ids_solution = ids(START, GOAL)
display_solution("ITERATIVE DEEPENING SEARCH (IDS)", ids_solution)
```

```python
# ============================================================
# DEMONSTRATION & EXECUTION: BFS
# ============================================================

def display_solution(algorithm_name, path):
    print(f"=== {algorithm_name} ===")
    if path is None:
        print("No solution found.")
        return

    print(f"Steps to Goal: {len(path) - 1}\n")
    for i, (state, move) in enumerate(path):
        if move:
            print(f"Step {i}: Move {move}")
        else:
            print("Initial State:")
        print_state(state)

# Run BFS
bfs_solution = bfs(START, GOAL)
display_solution("BREADTH FIRST SEARCH (BFS)", bfs_solution)
```
